# K. Max and Min

```python
inp = input()
nums = inp.split()

nums0 = int(nums[0])
nums1 = int(nums[1])
nums2 = int(nums[2])

max = nums0
if max < nums1:
    max = nums1
if max < nums2:
    max = nums2

min = nums0
if min > nums1:
    min = nums1
if min > nums2:
    min = nums2

print(min, max)
```

# C. Even, Odd, Positive and Negative

```python
size = int(input())
nums = input().split()

even = 0
odd = 0
positive = 0
negative = 0

for i in range(size):
    n = int(nums[i])
    if (n % 2 == 0):
        even = even + 1
    else:
        odd = odd + 1
    if n > 0:
        positive = positive + 1
    if n < 0:
        negative = negative + 1

print("Even:", even)
print("Odd:", odd)
print("Positive:", positive)
print("Negative:", negative)
```

# Q. Digits

```python
t = int(input())

for i in range(t):
    num = int(input())
    if num == 0:
        print(0)
        continue
    while num > 0:
        print(num % 10, end = " ")
        num //= 10
    print()
```
