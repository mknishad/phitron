# Tuples

```python
# declaration
tpl = (10, 20, 30, 40, 50)
print(type(tpl), tpl)

float_tpl = (10.1, 20.2, 30.3, 40.4, 50.5)
print(type(float_tpl), float_tpl)

mixed_tpl = (10, 20.2, 'abc', True)
print(mixed_tpl)

tpl1 = (10)
print(type(tpl1), tpl1)

lst = [10, 20, 30]
tpl = tuple(lst)
print(type(tpl), tpl)
```

```python
# Access
tpl = (1, 2.3, "str", True)

# Index
print(tpl[2])

# Slice
print(tpl[0:3])
```

```python
# Tuple Methods
tpl = (2, 3, 5, 5, 4, 3, 5, 1, 5)
print(tpl.count(5))                      
print(tpl.index(5))
```

```python
S = {1, 2, 3, 4, 2, 3}
print(S, type(S))
```

```python
# empty structures
B = []
print(type(B))

C = ()
print(type(C))

S = {}
print(type(S))

S = set()
print(type(S))
```

## Set Mathematical Methods

```python
a = {1, 3, 5, 7, 9}
b = {2, 4, 6, 8, 10}
c = {1, 3, 5, 7, 9, 11, 13, 15}
print(a.union(b))
print(a.intersection(b))
print(a.isdisjoint(b))
print(a.issubset(b))
print(a.issubset(c))
```

## Dicitonary

```python
d = {}
print(type(d))

d = {"one": 1, 1: "one"}
print(d["one"])
print(d[1])

d[1] = "uno"
print(d[1])

print(d.get(2))
print(d.get(2, "two"))

d.update({3: "three"})
print(d)

d2 = d.copy()
print(d2)

del d2[3]
print(d2)

d2.pop(1)
print(d2)

keys = d.keys()
values = d.values()
print(keys, values)

d["two"] = 2
print(keys, values)

items = d.items()
print(items)

for key, value in d.items():
    print(key, value)
```

## Dictionary Comprehension

```python
square = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print(square)
```

```python
co_ordinates = [(10, 10.5), (20.5, 192), (101, 102)]
locations = ["dhaka", "chattogram", "sylhet"]
exact_locations = {co_or: loc for co_or, loc in zip(co_ordinates, locations)}
print(exact_locations)
```

#### 2. Given a string, print the frequency of the words

```python
string = """ Mushfiqur Rahim marked his 100th Test with a masterclass in patience and poise, closing the opening day in Mirpur on an unbeaten 99 that carried Bangladesh to a strong 292 for 4 after an uncertain morning.
As he walked off just one run short of joining the elite list of players to score a century in their 100th Test on Wednesday, the Sher-e-Bangla crowd rose to salute a batter who had steadied the innings with the composure of a veteran and the hunger of a debutant.
The Tigers’ position owed much to Mushfiqur’s calm reconstruction job and to Mominul Haque’s valuable 63, as well as Litton Das’s fluent 47*. But the day had begun in Ireland’s favour, courtesy of Andy McBrine, who took all four wickets, giving away 82 runs.
Before a ball was bowled, Mushfiqur was greeted with a warm tribute: his family, former captains Akram Khan and Habibul Bashar, and the BCB marked the milestone with a presentation."""
words = string.split()
count = {}
for word in words:
    count[word] = count.get(word, 0) + 1
print(count)
```
