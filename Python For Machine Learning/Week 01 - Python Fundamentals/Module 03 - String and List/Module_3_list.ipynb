```python
numbers = [100, 30, 40, 50]
print(type(numbers))

new_list = numbers[0:3]
print(new_list)

new_list.append(80)
print(new_list)

new_list.insert(2, 90)
print(new_list)

print(new_list.pop())
print(new_list)

new_list.remove(100)
print(new_list)

everything = [2, 3.4, 'apple']
print(everything)

list_of_list = [[1, 10, 100], [2, 20, 200], [3, 30, 300]]
print(list_of_list)
```

```python
numbers.sort()
print(numbers)
```

```python
numbers.sort(reverse = True)
print(numbers)
```

## List as Stack

```python
stack = []
stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)

print(stack.pop())
print(stack.pop())
print(stack.pop())
print(stack[-1]) # top
```

## List as Queue

```python
queue = []
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)

print(queue[0]) # front
queue.pop(0)
print(queue[0])
```

```python
even = []

for i in range(101):
    if i % 2 == 0:
        even.append(i)

print(even)
```

```python
odd = [x for x in range(100) if x % 2 == 1]
print(odd)
```

```python
fruits = ['apple', 'orange', 'mango']
upper_fruits = [fruit.upper() for fruit in fruits]
print(upper_fruits)
```
