# String Methods

```python
string = "Welcome to Phitron ML course, by Phitron ML"
print(len(string))

lower_string = string.lower()
print(lower_string)

upper_string = string.upper()
print(upper_string)

capitalize_string = string.capitalize()
print(capitalize_string)

print("phitron" in string)

index = lower_string.find("phitron")
print(index)

reverse_index = lower_string.rfind("phitron")
print(reverse_index)

count = lower_string.count("phitron")
print(count)

replaced_string = string.replace("ML", "AI/ML")
print(replaced_string)
```

# Splitting, Joining, Formatting

```python
prompt = "What is token?"
tokens = prompt.split()
print(tokens)
print(type(tokens))

sentence = " ".join(tokens)
print(sentence)
print(type(sentence))

name = "Nishad"
age = 32
height = 5.77777
info = f"My name is {name.upper()}. I am {age} years old. I am {height:.2} feet tall."
print(info)

model_accuracy = 0.8454545
print(f"Model accuracy is {model_accuracy:.2%}")
```
