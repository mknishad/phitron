## E. Lowest Number

```python
n = int(input())
inp = input()
nums = inp.split()
a = []
for i in range(len(nums)):
    a.append(int(nums[i]))
min = a[0]
min_idx = 0
for i in range(len(a)):
    if a[i] < min:
        min = a[i]
        min_idx = i
print(min, min_idx + 1)
```

## M. Replace MinMax

```python
n = int(input())
inp = input()
nums = inp.split()

a = []
for i in range(n):
    a.append(int(nums[i]))

min = a[0]
min_idx = 0
max = a[0]
max_idx = 0

for i in range(n):
    if a[i] < min:
        min = a[i]
        min_idx = i
    if a[i] > max:
        max = a[i]
        max_idx = i

tmp = a[min_idx]
a[min_idx] = a[max_idx]
a[max_idx] = tmp

for i in range(n):
    print(a[i], end = ' ')
```

## K. Sum Digits

```python
n = int(input())
num = input()

sum = 0
for i in range(n):
    sum += int(num[i])
print(sum)
```

## H. Good or Bad

```python
t = int(input())

for i in range(t):
    s = input()
    if "010" in s:
        print("Good")
    elif "101" in s:
        print("Good")
    else:
        print("Bad")
```
